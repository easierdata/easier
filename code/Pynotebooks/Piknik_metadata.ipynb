```python
# inflate web3 information (df_sel) into stac.json
```

```python
# Initialize an empty dictionary to hold the result
result = {}

# Group the DataFrame by 'sceneid'
for sceneid, group_data in df_sel.groupby("sceneid"):
    # Initialize a dictionary for this 'sceneid'
    scene_dict = {}

    # Iterate through the group data
    for index, row in group_data.iterrows():
        filename, cid = row["filename"], row["FileCID"]

        # Initialize a list for this filename if it doesn't exist
        if filename not in scene_dict:
            scene_dict[filename] = []

        # Add the cid to the list for this filename
        scene_dict[filename].append(cid)

    # Add this scene_dict to the overall result
    result[sceneid] = scene_dict
```

```python
# loop to enrich stac


def enrich_web3(row):
    import json
    import os
    import urllib.parse

    input_stac_path = f"../../data/stac/landsat-c2l1/json/{str(row['AcqDate'])}/{str(row['sceneid'])}_stac.json"
    output_stac_path = f"../../data/stac/landsat-c2l1/web3_enriched_json/{str(row['sceneid'])}_stac.json"

    profile = result[row["sceneid"]]
    with open(input_stac_path) as input_file:
        stac_data = json.load(input_file)

    for asset_name, asset_data in stac_data["assets"].items():
        # print(asset_data)
        href = asset_data["href"]
        file_name = os.path.basename(urllib.parse.urlparse(href).path)
        if asset_name == "index" or "alternate" not in asset_data:
            if asset_name != "index":
                print("no alternate for", asset_name)
            continue
        if "IPFS" not in asset_data["alternate"]:
            asset_data["alternate"]["IPFS"] = {}
        # print(file_name)
        if file_name in profile:
            asset_data["alternate"]["IPFS"]["href"] = f"ipfs://{profile[file_name][0]}"
            # print(profile)
        else:
            print(f"FileCID not found for asset: {asset_name}")
    with open(output_stac_path, "w") as output_file:
        json.dump(stac_data, output_file, indent=4)
    return True
```

```python
# loop to enrich stac


def compute_cid(file_path):
    import subprocess

    cid = (
        subprocess.check_output(
            [ipfs_binary, "add", "-qn", "--cid-version", "1", file_path]
        )
        .decode()
        .strip()
    )
    return cid


def enrich_dual_web3(row):
    import json
    import os
    import urllib.parse
    import pathlib

    input_stac_path = f"../../data/stac/landsat-c2l1/json/{str(row['AcqDate'])}/{str(row['sceneid'])}_stac.json"
    output_stac_path = f"../../data/stac/landsat-c2l1/web3_enriched_json/{str(row['sceneid'])}_stac.json"

    profile = result[row["sceneid"]]

    with open(input_stac_path) as input_file:
        stac_data = json.load(input_file)

    for asset_name, asset_data in stac_data["assets"].items():
        # print(asset_data)
        href = asset_data["href"]
        file_name = os.path.basename(urllib.parse.urlparse(href).path)
        file_path = f"../../data/landsat/piknik/{str(file_name)}"
        if asset_name == "index" or "alternate" not in asset_data:
            if asset_name != "index":
                print("no alternate for", asset_name)
            continue
        if pathlib.Path(file_path).is_file():
            cid = compute_cid(file_path)
            if "IPFS" not in asset_data["alternate"]:
                asset_data["alternate"]["IPFS"] = {}
            asset_data["alternate"]["IPFS"]["href"] = f"ipfs://{cid}"
        else:
            print(f"File not found for asset: {asset_name}")

        if "Filecoin" not in asset_data["alternate"]:
            asset_data["alternate"]["Filecoin"] = {}
        # print(file_name)
        if file_name in profile:
            asset_data["alternate"]["Filecoin"][
                "href"
            ] = f"ipfs://{profile[file_name][0]}"
            asset_data["alternate"]["Filecoin"]["storage:platform"] = "PiKNiK"
            # print(profile)
        else:
            print(f"Filecoin CID not found for asset: {asset_name}")
    with open(output_stac_path, "w") as output_file:
        json.dump(stac_data, output_file, indent=4)
    return True
```
